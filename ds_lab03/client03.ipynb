```python
import xmlrpc.client
```

```python
host = "127.0.0.1"
```

```python
port = 65003
```

```python
def main():
    proxy = xmlrpc.client.ServerProxy(f"http://{host}:{port}/")
    print("Calling remote add(4,7) ->",proxy.add(4,7))
    print("Calling remote factorial(5) ->",proxy.factorial(5))
    print("Calling remote reverse_string('distributed') ->",proxy.reverse_string("distributed"))
```

```python
if __name__ == "__main__":
    main()
```
