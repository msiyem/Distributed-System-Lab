```python
import socket
import threading
import time
```

```python
HOST = "127.0.0.1"
PORT = 65005
```

```python
N = 5
```

```python
def worker(i):
    with socket.socket(socket.AF_INET, socket.SOCK_STREAM) as c:
        c.connect((HOST, PORT))
        c.sendall(f"reques-{i}".encode())

        reply = c.recv(1024).decode()
        print(f"[CLIENT {i}] {reply}")
```

```python
def main():
    start = time.time()

    threads = [threading.Thread(
        target=worker,
        args=(i,)
    ) for i in range(N)]

    for t in threads:
        t.start()
    for t in threads:
        t.join()
    elapsed = time.time() - start

    print(f"\nAll {N} requests completed in {elapsed:.2f} seconds")
    print(f"(with Max_Workers = 2 on the server, expect roughly 6s, not ~2s or ~10s)")
```

```python
if __name__ == "__main__":
    main()
```
