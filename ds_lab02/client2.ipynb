```python
import socket
```

```python
import threading
```

```python
import time
```

```python
host = "127.0.0.1"
port = 65002
```

```python
n = 50
```

```python
def worker(i):
    with socket.socket(socket.AF_INET, socket.SOCK_STREAM) as s:
        s.connect((host,port))
        s.sendall(f"request-{i}".encode())
        reply = s.recv(1024).decode()
        print(f"client {i} {reply!r}")
```

```python
def main():
    start = time.time()
    threads = [threading.Thread(
        target = worker,
        args = (i,)) for i in range(n)
    ]
    for t in threads:
        t.start()
    for t in threads:
        t.join()
    elapsed = time.time() - start
    print(f"\n All {n} requests complete in {elapsed:.2f} seconds")
    print("(with a thread server this should be ~2s, not ~{}s)".format(n*2))
    
    
```

```python
if __name__ == "__main__":
    main()
```
