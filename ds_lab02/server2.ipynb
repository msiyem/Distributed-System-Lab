```python
import socket
```

```python
import threading
```

```python
import time
```

```python
host = "127.0.0.1"
port = 65002
```

```python
def handle_client(conn,addr):
    with conn:
        print(f"[worker - {threading.get_ident()}] handing {addr}")
        request = conn.recv(1024).decode()
        time.sleep(2)
        reply = f"Processed {request!r} by worker thread {threading.get_ident()}"
        conn.sendall(reply.encode())
        print(f"[worker - {threading.get_ident()!r}] done with {addr}")
```

```python
def main():
    with socket.socket(socket.AF_INET, socket.SOCK_STREAM) as s:
        s.setsockopt(socket.SOL_SOCKET,socket.SO_REUSEADDR,1)
        s.bind((host,port))
        s.listen(50)
        print(f"[Dispather] listening on {host}:{port!r}")
        while True:
            conn,addr = s.accept()
            print(f"[Dispather] accepted {addr} , spawing worker thread")
            worker = threading.Thread(target = handle_client, args = (conn,addr))
            worker.start()
```

```python
if __name__ == "__main__":
    main()
```
