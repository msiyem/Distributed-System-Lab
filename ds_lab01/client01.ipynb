```python
import socket
```

```python
host = "127.0.0.1"
port = 65001
```

```python
def main():
    with socket.socket(socket.AF_INET,socket.SOCK_STREAM) as c:
        c.connect((host,port))
        print(f"client is connect with {host}:{port}")
        msg = input(">")
        print(f"client send message: {msg!r}")
        c.sendall(msg.encode())

        reply = c.recv(1024).decode()

        print(f"client receive message: {reply!r}")
        
        
    
```

```python
if __name__ == "__main__":
    main()
```
