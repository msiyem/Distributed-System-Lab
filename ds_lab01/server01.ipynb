```python
import socket
```

```python
host = "127.0.0.1"
```

```python
port = 65001
```

```python
def main():
    with socket.socket(socket.AF_INET,socket.SOCK_STREAM) as s:
        s.bind((host,port))
        s.listen(5)
        print(f"server is listenning on {host}:{port} and wait for client....")

        while True:
            conn,addr = s.accept();
            print(f"server is connect with client: {addr!r}")
            with conn:
                req = conn.recv(1024).decode();
                print(f"server receive: {req!r}")
                
                rep = req.upper()
                
                print(f"server send: {rep!r}")
                conn.sendall(rep.encode())

                
                
            
        
    
    
```

```python
if __name__ == "__main__":
    main()
```
